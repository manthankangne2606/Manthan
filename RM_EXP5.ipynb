```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("fake reviews dataset.csv")

print(df.head())
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicates:")
print(df.duplicated().sum())
```

```python
print(df["label"].value_counts())
```

```python
sns.countplot(x="label", data=df)

plt.title("Distribution of Fake and Genuine Reviews")
plt.xlabel("Review Type")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
df["Review_Length"] = df["text_"].str.len()
```

```python
df["Word_Count"] = df["text_"].str.split().str.len()
```

```python
df["Exclamation_Count"] = df["text_"].str.count("!")
```

```python
print(df[[
    "text_",
    "Review_Length",
    "Word_Count",
    "Exclamation_Count"
]].head())
```

```python
sns.countplot(x="label", data=df)

plt.title("Fake vs Genuine Reviews")
plt.xlabel("Review Type")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
plt.hist(df["Review_Length"], bins=30)

plt.xlabel("Review Length")
plt.ylabel("Number of Reviews")
plt.title("Distribution of Review Length")
plt.show()
```

```python
sns.countplot(x="rating", data=df)

plt.title("Distribution of Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
sns.boxplot(x="label", y="Review_Length", data=df)

plt.title("Review Length by Review Type")
plt.xlabel("Review Type")
plt.ylabel("Review Length")
plt.show()
```
